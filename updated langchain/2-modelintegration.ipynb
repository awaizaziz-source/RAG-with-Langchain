MODEL INTEGRATION WITH OPENAI,GROQ AND GOOGLE GEMINI

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

GROQ MODEL INTEGRATION


```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-20b")
response = model.invoke("hello, why do parrots talk?")
response
```

STREAMING

```python
##in streaming we get response right after giving the input to it , and it will keep on streaming the response until it is completed. This is useful when we want to get the response as soon as possible and not wait for the entire response to be generated.
for chunk in model.stream("Write a 100 words paragraph on a girl struggling through life."):
    if chunk.content:
        print(chunk.content, end="", flush=True)
```

```python
##batch:giving inputs to the LLM in parallel and getting outputs in parallel
responses=model.batch([
    "why do parrots have colorful feathers?"
    "is coding difficult for a girl?"
    "give me a motivational sentence."
                       
])
for response in responses:
    print(response.content)
```
